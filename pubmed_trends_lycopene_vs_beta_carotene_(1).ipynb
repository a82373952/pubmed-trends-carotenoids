# Analiza trendów publikacji w PubMed
### Zapytania:
- `lycopene[Title/Abstract]`
- `beta carotene[Title/Abstract]`

Zakres lat: 2000–2025

```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
data_lycopene = {
    'Year': [2025,2024,2023,2022,2021,2020,2019,2018,2017,2016,2015,2014,2013,2012,2011,2010,
             2009,2008,2007,2006,2005,2004,2003,2002,2001,2000],
    'Count': [346,412,415,481,45,388,351,32,319,317,3,283,282,297,258,229,
              218,212,227,194,19,166,155,15,128,142]
}

data_beta = {
    'Year': [2025,2024,2023,2022,2021,2020,2019,2018,2017,2016,2015,2014,2013,2012,2011,2010,
             2009,2008,2007,2006,2005,2004,2003,2002,2001,2000],
    'Count': [58,76,792,865,794,769,762,659,688,606,602,639,62,631,574,532,
              509,475,464,464,461,451,432,381,422,423]
}

df_lycopene = pd.DataFrame(data_lycopene)
df_beta = pd.DataFrame(data_beta)

df_merged = pd.merge(df_lycopene, df_beta, on='Year', how='outer', suffixes=('_Lycopene', '_BetaCarotene'))
df_merged = df_merged.sort_values('Year')
df_merged
```

```python
# Wykres trendów
plt.figure(figsize=(12,6))
plt.plot(df_merged['Year'], df_merged['Count_Lycopene'], marker='o', label='Lycopene')
plt.plot(df_merged['Year'], df_merged['Count_BetaCarotene'], marker='o', label='Beta carotene')
plt.xlabel('Rok')
plt.ylabel('Liczba publikacji')
plt.title('Trendy publikacji w PubMed (2000–2025)')
plt.legend()
plt.grid(True)
plt.show()
```

### Wnioski
- **Beta-karoten**: temat bardzo intensywnie badany, liczba publikacji w latach 2000–2025 jest wielokrotnie wyższa niż dla likopenu.
- **Lycopene**: także szeroko badany, ale na mniejszą skalę; widoczny wzrost liczby publikacji w ostatnich latach.
- Wyniki pokazują, że oba karotenoidy są istotne naukowo, ale beta-karoten ma znacznie dłuższą i bardziej rozbudowaną historię badań.
